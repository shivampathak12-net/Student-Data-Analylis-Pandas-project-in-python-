```python
import pandas as pd
student = {
    'Name' : ['Rahul','Priya','Aman','Sneha','Karan'],
    'Age' : [20,21,19,20,22],
    'Marks' : [85,92,67,78,56]
}
df_students = pd.DataFrame(student)
print("====== Your student data ======")
print(df_students)
print("\n====== Head of student data ======")
print(df_students.head(3))
print("\n====== Average Marks =======")
print("Average marks : ",df_students['Marks'].mean())
print("\n ===== Students scoring above 75! ========")
print(df_students[df_students['Marks'] >= 75])
print("\n===== Sorted by Marks ======")
print(df_students.sort_values(by='Marks',ascending=False))
```
